# OOP in Python
Un oggetto in Python ha tre caratteristiche:
 - **identità**, distingue un oggetto dall'altra anche se hanno gli stessi dati. In altre parole con identità identifichiamo la posizione in memoria di un oggetto che possiamo calcolare con la funzione id(oggetto)
 - **stato**, i dati che rappresenta in un certo momento e definiti tramite gli *attributi*
 - **comportamento** i *metodi* che può eseguire


```python
class Cane:
    def __init__(self, nome, eta):
        self.nome = nome
        self.eta = eta

    def verso(self):
        print("BAU")


fido = Cane("Fido", 3)
print(fido.nome)
print(fido.eta)
fido.verso()


fido2 = Cane("Fido", 3)

print(id(fido2))
print(id(fido))
```

## Incapusalmento
Come per altri linguaggi OOP, anche Python offre delle possibilità per gestire l'incapsulamento, nello specifico vengono usate le convenzioni di scrittura:

 - **Pubblico: variabile**
 - **Protetto: _variabile**
 - **Privato: __variabile**

L'incapsulamento è utile per definire regole di sicurezza ai dati

```python
class Conto:
    def __init__(self, saldo):
        self.__saldo = saldo

    def aggiungi_saldo(self, importo):
        self.__saldo = self.__saldo + importo

    def stampa_saldo(self):
        print(self.__saldo)


c = Conto(2000) 
c.aggiungi_saldo(300)
c.stampa_saldo()

# c.__saldo NON E' ACCESSIBILE ALL'ESTERNO
```

# Generalizzazione
La **generalizzazione** consiste nell'individuare caratteristiche comuni di più classi e portarle in una classe più generale o **astratta**. Questa permette di:

- Ridurre la duplicazione
- Favorire la riusabilita 
- Organizzare le classi in una gerarchia logica


```python
class Persona():
    def __init__(self, nome):
        self.nome = nome

class Studente(Persona):
    pass

class Docente(Persona):
    pass



p = Persona("Ciccio")
print(p.nome)


s = Studente("Franco")
print(s.nome)


d = Docente("Lavinia")
print(d.nome)
```

In altre parole Docente e Studente ereditano la struttura della classe Persona

## Metodi di istanza, classe e statici


```python
class Persona():
    numero_persone = 0 #attributo di classe

    #Questi sono metodi di istanza in quanto variano da istanza a istanza
    def __init__(self, nome):
        self.nome = nome
    def get_nome(self):
        return self.nome

    #Questi sono invece i metodi di class utili a condividere variabili tra istanze di una stessa classe
    @classmethod #decoratore per definire metodi di classe
    def nuova_persona(cls): #prende il modificatore cls per accedere agli attributi di classe
        cls.numero_persone = cls.numero_persone + 1

    @staticmethod
    def specie():
        return "Umano - Mammifero"


p = Persona("Ciccio")
p.nuova_persona()
print(p.get_nome())

p2 = Persona("Franca")
p2.nuova_persona()
print(p2.get_nome())

print(Persona.numero_persone)  #possiamo accedere ai metodi di classe in questo modo
print(Persona.specie())


```

## Metodi speciali __init__ e __str__ e altri..

I primi due metodi li abbiamo già affrontati in qualche esercizio precedente, invece *__dpr__* è un nuovo metodo. Questo si differenzia rispetto __str__ semplicemente per la convenzione che se ne fa infatti:

- __str__ è un metodo toString() utile a chi usa il programma
- __dpr__ è un metodo toString() ma per usi più tecnici, tipo stampa di informazioni di log

```python
class Persona():
    def __init__(self, nome, cognome):
        self.nome = nome
        self.cognome = cognome

    def __str__(self):
        return f"{self.nome} {self.cognome}"
    
    def __dpr__(self):
        return f"Persona({self.nome},{self.cognome})"

p = Persona("Patrick", "Nicolosi")
print(p.__str__())
print(p.__dpr__())
```

Altri metodi sono:

-__len__ -> definisce il comportamento della funzione len() su quell'oggetto
-__eq__ -> definisce il comportamento dell'operatore == su quell'oggetto
-__add__ -> definisce il comportamento dell'operatore + su quell'oggetto
-__getitem__ -> definisce il comportamento dell'operatore di accesso [] su quell'oggetto

## Ereditarietà 

Di base per ereditarietà intendiamo quel meccanismo della OOP secondo il quale possiamo definire le caratteristiche di una classe partendo da un'altra classe detta madre. 

```python
#Esempio

class Veicolo():
    def __init__(self, nome):
        self.nome = nome

class Auto(Veicolo):
    pass                    #eredita tutto ciò che è presente in veicolo
```

```python
#Esempio di overridding

class Veicolo():
    def __init__(self, marca):
        self.marca = marca

    def descrivi(self):
        print(f"Sono un Veicolo di marca {self.marca}")

class Auto(Veicolo):

    def descrivi(self):
        #super().descrivi()      --> qualora intendessimo richiamare anche Veicolo.descrivi()
        print(f"Sono un'Auto di marca {self.marca}")



v = Veicolo("fiat")
v.descrivi()    

a = Auto("ferrari")
a.descrivi()          
```

```python
#Esempio


class Animale():
    def __init__(self, nome):
        self.nome = nome

    def verso(self):
        print(f"Non sono sicuro sul verso di {self.nome}")

class Cane(Animale):
    def verso(self):
        print(f"{self.nome} fa BAU")

class Gatto(Animale):
    def verso(self):
        print(f"{self.nome} fa MEOW")

a = Animale("Dumbo")
a.verso()

b = Cane("Fido")
b.verso()

c = Gatto("Spoki")
c.verso()
```

Con l'overridding del metodo abbiamo involontariamente introdotto il concetto di **polimorfismo**, dove per polimorfismo si intende:

*Il polimorfismo è uno dei concetti chiave della programmazione orientata agli oggetti (OOP). Il termine significa "molte forme" e indica la capacità di usare lo stesso metodo o operatore su oggetti diversi, ottenendo comportamenti coerenti ma specifici per ogni tipo. In altre parole, diverse classi possono avere metodi con lo stesso nome, ma ognuno agirà secondo le proprie caratteristiche.*

Un'altra forma di polimorfismo lo si ha quando effettuiamo override di operatori tipo

```python
print([1] + [2]) #[1,2]
```

```python
#esempio override operatore +

class Punto:
    def __init__(self, x, y):
        self.x = x 
        self.y = y

    def __str__(self):
        return f"({self.x},{self.y})"

    def __add__(self, other):
        return Punto(self.x + other.x, self.y + other.y)

p = Punto(12, 4)
p2 = Punto(34, 2)

print(p + p2) 
```

```python
#Crea una classe base Forma con metodo area().
#Crea due classi derivate:
# - Rettangolo -> area = base * altezza
# - Cerchio -> area = pi * r^2
#Crea una lista di forme e stampa l'area di ciascuna



class Forma:
    def area(self):
        pass

class Rettangolo(Forma):
    def __init__(self, base, altezza):
        self.base = base
        self.altezza = altezza

    def area(self):
        return self.base * self.altezza


class Cerchio(Forma):
    pi = 3.14

    def __init__(self, r):
        self.r = r

    def area(self):
        return self.pi * (self.r ** 2)


l = [Cerchio(12), Rettangolo(2, 3), Cerchio(4), Rettangolo(2,2)]

for i in range(len(l)):
    print(f"L'area in questione è {l[i].area()}")
```

## Classi astratte
Il concetto di classe astratta è il medesimo di tutti gli altri linguaggi: forniscono un contratto a cui legarsi, per cui tutte le classi legate ad una astratta dovranno rispettare quel contratto.

In python la gestione delle classi astratte avviene tramite il modulo **abc** *(abstract base classes)*

In particolare, per rendere una classe astratta, dobbiamo:
1. Importare *abc* e *abstractmethod*
2. Una classe diventa astratta se eredita da *ABC*
3. Un metodo diventa astratto se decorato con **@abstractmethod**



**UNA CLASSE ASTRATTA NON SI PUO ISTANZIARE**

```python
from abc import ABC, abstractmethod

class Animale(ABC):
    @abstractmethod
    def verso(self):
        pass

class Cane(Animale):        #implementazione corretta della classe astratta Animale
    def verso(self):
        print("BAU")

class Gatto(Animale):
    pass

#a = Animale()  
                            #Restituisce errore
c = Cane()                  #Questo invece funziona!
c.verso()



g = Gatto()                 #Questo non funziona! infatti in fase runtime Python controlla come sono definite le classi e questa non è stata definita correttamente in quanto manca la definzione del metodo verso() -> questo a runtime e no compiletime      
```

```python
#Dichiaro solo un metodo abstract su due

from abc import ABC, abstractmethod

class Animale(ABC):
    def razza(self):
        pass

    @abstractmethod
    def verso(self):
        pass

class Cane(Animale):       
    def verso(self):
        print("BAU")

c = Cane()   #Anche se non ho definito in Cane il metodo razza, non ho comunque problemi perche non dichiarato come abstract
c.verso()



```

## Decoratori
In Python esistono diversi decoratori, alcuni li abbiamo già incontrati:

- *@staticmethod* --> serve a definire un metodo statico, non necessita del parametro self. Si usa per definire tutti quei metodi di utilità che non hanno necessità di accedere a dati o oggetti della classe.
- *@classmethod* --> serve a definire un metodo legato alla classe, necessita del parametro cls. A differenza dei primi, con i quali condividono il modo in cui vengono chiamati, accedono a dati della classe e ne modificano il valore.
- *@property* --> permette di trattare un metodo come fosse un attributo
- *@nome.setter* --> permette di definire un metodo setter pe un attributo


```python
#staticmethod

class MathUtils():
    @staticmethod
    def sum(a, b):
        return a + b

print(MathUtils.sum(2, 43))
```

```python
#classMethod
class Object:
    __counter = 0

    @classmethod
    def get_objs_number(cls):
        return cls.__counter

    @classmethod
    def add_obj(cls):
        cls.__counter = cls.__counter + 1


o1 = Object()
Object.add_obj()

o2 = Object()
Object.add_obj()

o3 = Object()
Object.add_obj()


print(Object.get_objs_number())


#Come vediamo add_obj() modifica un attributo di classe che è __counter
```

```python
#classmethod 
#un altro caso in cui vengono usati questi metodi è per costruire oggetti usando costruttori alternativi

class Person():
    def __init__(self, name, surname):
        self.name = name
        self.surname = surname

    @classmethod
    def from_string(cls, string):
        splitted = string.split(" ") 
        return cls(splitted[0], splitted[1])

    def __str__(self):
        return f"{self.name} {self.surname}"

p = Person("Patrick", "Nicolosi")
print(p)

p2 = Person.from_string("Patrick Nicolosi")
print(p2)
```

```python
#property

class Rect:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    @property
    def area(self):
        return self.x * self.y

r = Rect(12, 3)

print(r.area)   #cosi OK!
print(r.area()) #se provo ad accedere al metodo usando la classica notazione mi da errore!
```

```python
# @nome.setter
# Permette di costruire sistemi di validazione sui dati

class Person:
    def __init__(self, name):
        self.name = name

    @property
    def name(self):
        return self._name

    @name.setter
    def name(self, name):
        if name.isalpha():
            self._name = name
        else:
            print("Devi inserire un testo valido")


p = Person("Ciccio")
print(p.name)  # Ciccio

p.name = "123"  # Devi inserire un testo valido
print(p.name)   # Ciccio
```

```python
#Esercizio
#Crea una classe Studente che abbia:
#@classmethod per creare uno studente a partire da una stringa tipo "Luca-20-Matematica"
#@property per calcolare automaticamente l'anno di nascita dall'età
#@property con setter per impedire età negative


class Student():
    def __init__(self, name, age, subject):
        self.name = name
        self._age = age
        self.subject = subject

    @property
    def birthday_year(self):
        return 2026 - self._age

    @classmethod
    def from_str(cls, str):
        spl = str.split("-")
        return cls(spl[0], spl[1], spl[2])

    @property
    def age(self):
        return self._age

    @age.setter
    def age(self, age):
        if age < 0:
            print("The age has to have positive")
        else:
            self._age = age


    def __str__(self):
        return f"{self.name}-{self._age}-{self.subject}"

s = Student("Patrick", 27, "Informatics")
print(s.birthday_year)
print(s)

s2 = Student.from_str(s.__str__())
print(s2)

s2.age = -24



```
