# Casting tra tipi
Esistono due tipi di casting:
- **implicito**, gestito in maniera automatica da Python
- **esplicito**, gestito da chi scrive il codice tramite funzioni come *float(), int(), str()*. In generale il casting esplicito ci da più controllo e quindi è preferito rispetto al primo tipo

`````python
#Esempio di conversione esplicita
a = 10
b = float(a)

print(type(b))
print(b)
`````

`````python
#Esempio di conversione implicita
a = 10
b = 10 + 50.3

print(type(b))
print(b)
`````

### Boolean Casting

Esiste un tipo particolare di conversione detta **boolean casting** che converte un dato in un booleano. In particolare

- *bool(0)* ritorna *False*
- *bool(12)* ritorna *True*
- *bool("Ciao")* ritorna *True*
- *bool("")* ritorna *False*

La logica alla base di questa conversione è "se quello passato come parametro della funzione bool è qualcosa di 'vuoto' e quindi una stringa vuota o 0 allora ritorna *False*, altrimenti *True*.

## Errori comuni tra casting

Non tutti i cast sono possibili, ad esempio fare *int("abc")* ci restituirà un errore **Value Error**.
In questi casi è sempre buona norma validare l'input oppure fare un controllo tramite blocco *try/except*

`````python
#Esempio di errore di conversione
a = input("Quanti anni hai?")
print(a + 5)

#Questo codice darà errore in quanto il tipo di ritorno della funzione input è una stringa e quindi Python tenterà di fare un casting implicito a int con a + 5 che fallirà
`````

`````python
a = input("Quanti anni hai?")
print(int(a) + 5)

#Questo codice funzionerà in quanto abbiamo fatto un casting esplicito
`````

`````python
# un modo per gestire l'errore di conversione è quello di usare un blocco try-except per catturare l'eccezione ValueError che viene sollevata quando si tenta di convertire una stringa non numerica in un intero.
a = input("Quanti anni hai?")

try: 
    print(a + 5)

except ValueError:
    print("Errore: non puoi sommare una stringa a un intero. Per favore inserisci un numero valido.")
`````

`````python
#Esercizio
# Chiedi all'utente di inserire un numero intero
# Converti questo numero in un float e stampalo
# Converta lo stesso numero in una stringa e stampalo insieme ad un messaggio

n = int(input("Inserisci un numero intero: "))
n_float = float(n)

print("Il numero convertito è", str(n_float))

`````

# Le Stringhe

## Operazioni tra stringhe

Una stringa è una sequenza di caratteri e come tutte le sequenze possiamo manipolarle attraverso alcuni operatori. Il primo operatore che vediamo è l'operatore di accesso *[indice]*. Eseguendo un codice del tipo

```python
s = "ciao vita mia"
print(s[0])
```
Otteniamo come output 

```python
c
```

in quanto l'accesso [0] ci permette di accedere al primo carattere della stringa e quindi la 'c'. L'operatore di accesso può anche accettare sintassi un po piu complesse per effettuare operazioni di accesso specifiche. Ad esempio

```python
s = "ciao vita"
print(s[0:2])
```
ci permette di effettuare il cosiddetto **slicing** di una stringa, in particolare il risultato che otterremo sarà 

```python
ci
```

in quanto nello slicing il secondo indice (in questo caso il 2) è sempre escluso, mentre il primo incluso.

Altre operazioni svolgibili con le stringhe sono le seguenti:
- **concatenazione** : "ciao" + "mondo" = "ciaomondo"
- **ripetizione** : "ha" * 5 = "hahahahaha"
- **lunghezza** : len("Python") = 6
- **accesso** : "Python"[0] = "P"
- **slicing**: "Python"[1:4] = "yth"

Altre funzioni comuni per manipolare le stringhe in python sono:
-  ```upper()```, ```lower()```, ```capitalize()```
-  ```strip()```, utile a togliere spazi all'interno delle stringhe
- ```replace(old, new)```per sostituire tutte le occorrenze del carattere di *old* con *new*
- ```split(sep)```
- ```join(iterable)```
- ```find()```
- ```index()```
- ```count(carattere)``` ritorna un int e conta il numero di occorrenze di un carattere
- ```startsWith(carattere)``` ritorna un bool è controlla se la stringa ha come primo carattere quello passato come argomento 'carattere'
- ```endsWith(carattere)```


## Possibili errori su stringhe
Uno degli errori più comuni quando manipoliamo stringhe è **IndexError** che si presenta ogni volta che noi tentiamo di accedere ad una posizione della stringa che non esiste, ad esempio:

```python
"ciao"[10] #IndexError
```

In questo caso darà errore perchè stiamo tentando di accedere alla posizione 10 di una stringa "ciao" che non ha questo numero di caratteri. Cosa che invece ci viene permessa da Python è passare all'operatore di accesso una stringa negativa non ottenendo errori.

```python
"ciao"[-1] #Stamperà il carattere 'o'
```
Quindi la logica che ci sta dietro l'indexing negativo è "dato un indice n, torna indietro di n passi partendo dal primo carattere della stringa"

L'accesso alla lista avviene tramite l'operatore di accesso parentesi quadra [], Python offre delle funzionalità interessanti come l'uso degli indici negativi, per cui per effettuare ad esempio l'accesso all'ultimo elemento di una lista basterebbe fare
```lista[-1]``` molto meno prolisso rispetto ad altri programmi in cui avremmo dovuto fare una cosa del genere ```lista[len(lista) - 1]```.

# Tuple
Sono un altro tipo di collezione rispetto le liste e le stringhe di tipo immutabile. La dichiarazione delle tuple avviene attraverso la seguente sintassi

```python
wd = ('Lun', 'Mar', 'Mer', 'Gio', 'Ven', 'Sab', 'Dom')
```

Alla tupla possiamo sicuramente accedere tramite operatore parentesi quadra e quindi

```python

```

Ma non possiamo sostituirne un valore perchè appunto **immutabile**, per cui

Possiamo iterare sulle tuple così come per le liste e quindi 

```python

```

Le operazioni principali su una tupla sono:

- ```len(tuple)``` ritorna la lunghezza della tupla
- ```t1 + t2``` concatena t2 a t1 come per le stringhe
- ```t[i1:i2]`` effettua lo **slicing** della tupla da i1 a i2 escluso
- ```t.count(value)``` conta il numero di occorrenze di value
- ```t.index(value)``` ritorna l'indice di value se questo esiste all'interno della tupla t

# Set
I set sono una collezione di oggetti **non ordinata** che non ammette duplicati (esattamente il cocetto di insieme in matematica).

```python
fruits = {'apple', 'banana', 'kiwi'}
```

Come visto per dichiare un set usiamo le parentesi graffe, ma esistono altri modi per creare un set, uno di questi è a partire da una lista

```python
fruits = set(['apple', 'banana', 'kiwi', 'banana'])
```

**Attenzione** quando creiamo un set da una lista rimuoviamo praticamente i duplicati dalla lista, questo potrebbe essere un buon modo per "ripulire" una lista.

Per cercare un elemento come per gli altri tipi di collezione possiamo usare l'operatore **in**, quindi

```python
fruits = {'apple', 'banana', 'kiwi'}
print('apple' in fruits) #Stampa True
```

Questa ricerca è molto rapida, proprio perchè i set usano delle strutture interne a Python ottimizzazione per avere un accesso veloce.

Per aggiungere elementi possiamo usare il metodo ```s.add(element)``` oppure ```s.update([e1, e2])``` per aggiungere più elementi in una volta.

Possiamo rimuovere elementi in due modi differenti:
- ```s.discard(e)``` che rimuove senza dare errore se l'elemento non è presente
- ```s.remove(e)``` che rimuove dando errori se l'elemento non è presente

`````python
s = {'patrick', 'karola', 'william', 'katya'}
s.discard('katya')

s.remove('katya')
`````

## Operazioni insiemistiche
Come detto i set sono dei veri e proprio insiemi matematici e quindi su questi possiamo fare delle operazioni particolari di :

- **unione**
- **intersezione**
- **differenza**
- **differenza simmetrica**

`````python
A = {1, 2, 3, 4, 5}
B = {6, 7, 8, 9, 1}

print(A.intersection(B))          #Prende gli elementi che stanno in entrambi gli insiemi
print(A.union(B))                 #Prende gli elementi sia di A che B
print(A.difference(B))            #Prende solo gli elementi di A
print(A.symmetric_difference(B))  #Prende solo gli elementi che appartengono ad A o B e non alla loro intersezione
`````

`````python
#Una notazione alternativa per manipolare gli insiemi

A = {1, 2, 3, 4, 5}
B = {6, 7, 8, 9, 1}

print(A & B)    #Prende gli elementi che stanno in entrambi gli insiemi
print(A | B)    #Prende gli elementi sia di A che B
print(A - B)    #Prende solo gli elementi di A
print(A ^ B)    #Prende solo gli elementi che appartengono ad A o B e non alla loro intersezione
`````

# Dizionari
Si tratta di una struttura dati molto potente che associa valori dinamici a chiavi sempre di tipo string. Di seguito una possibile sintassi utile per dichiarare un Dizionario.

```python
p = {"Patrick":26, "Serena":27}
```

Un altro modo per dichiarare un dizionario è

```python
p = dict(nome = "Luca", eta = 25)
```

Il valore di un dizionario è sempre accedibile tramite operatore [], per cui per il dict ```p = {"Patrick":26, "Serena":27}``` possiamo accedere all'età di Serena con la sintassi

```python
p["Serena"]
```

Possiamo modificare il valore di una chiave semplicemente assegnandone un nuovo valore

```python
p["Serena"] = 26
```

Per eliminare un elemento, possiamo usare ```del``` attraverso la sintassi

```python
del(p["Serena"])
```

Per svuotare un dizionare possiamo usare il metodo ```p.clear()```.
Possiamo controllare se una chiave è presente nel dizionario semplicemente attraverso l'operatore ```in```, quindi

```python
if "Serena" in p:
    print("è presente!")
```
Cercare all'interno dei dizionari attraverso l'operatore in è molto rapido grazie alla loro implementazione a basso livello attraverso Hash Table.

Per unire un dizionario ad un altro possiamo usare il metodo ```p.update(p2)```.

`````python
p = {"Patrick":26, "Serena":27}
print(p)

p["Serena"] = 26
print(p)

if "Serena" in p:
    print("è presente!")

del(p["Serena"])
print(p)

p.clear()
print(p)
`````

## Metodi di lettura nei dict
Abbiamo tre principali modi per leggere i dizionari (operatore di accesso [] a parte), ed in particolare

- **per chiave**, attraverso il metodo ```p.keys()```
```python
for k in p.keys():
    print(k)
```

- **per valore**, attraverso il metodo ```p.values()```
```python
for v in p.values():
    print(v)
```

- **per chiave,valore insieme**, attraverso il metodo ```p.items()```
```python
for k,v in p.items():
    print(f"{k} : {v}")
```

`````python
p = {"Patrick":26, "Serena":27}

for k in p.keys():
    print(k)

print("----")

for v in p.values():
    print(v)

print("----")

for k,v in p.items():
    print(f"{k} : {v}")
`````

# Funzioni
Per definire una funzione usiamo la keyword ```def``` quindi come esempio

```python
#Per definirla
def saluta():
    print("Ciao")

#Per richiamarla
saluta()
```

Una funzione può anche **ritornare un valore** e quindi in questo caso avremmo

```python
def square(x):
    return x*x
```


Una funzione è parte di codice riutilizzabile, non solo da noi stessi, quindi una parte fondamentale è documentare la struttura di una funzione, per questo esistono le **docstring**, l'utilizzo è il seguente

`````python
#uso delle docstring
def moltiplica(x):
    """Questa funzione calcola il quadrato di un numero"""
    return x*x

help(moltiplica)
`````

Una **funzione anonima** è una funzione definita inline e senza nome. Per definire una funzione anonima usiamo la keyword **lambda**

```python
print((lambda x,y:x+y)(2,3)) #funzione lambda che fa la somma
```

Queste funzioni sono molto usate con metodi come ```map, filter, sorted``` ecc

Di seguito una serie di esempi sul **passaggio di parametri**.

`````python
#passaggio di parametri per posizione
def area_rettangolo(base, altezza):
    return base*altezza

print(area_rettangolo(12, 34))
`````

`````python
#Passaggio di parametri nominali e con valori di default
def saluta(nome, saluto = "Ciao"):
    print(f"{saluto} {nome}")

saluta(nome="Patrick")
saluta(nome="Patrick", saluto="Salve")
`````

`````python
#Passaggio di argomenti tramite *args

def somma(*args):
    print(type(args)) #ritornerà tipo tupla
    return sum(args)

print(somma(12, 12, 3))

#Si tratta di un modo per passare argomenti come tupla
`````

`````python
def saluta(*args):
    print(f"{args[0][0]}, {args[0][1]}")

t = ('Patrick', 12)
saluta(t)
`````

## Esercizi sulle funzioni

`````python
#Esercizio
# Simulare una lista di contatti e quindi prevedere le seguenti azioni: aggiungere un contatto, eliminare un contatto, aggiornare un contatto, cercare un contatto per nome, visualizzare i contatti in ordine alfabetico

def add_contact(contact, contacts):
    contacts.append(contact)
    print(f"Aggiunto {contact[0]}")

def remove_contact(contact, contacts):
    removed = contacts.pop(contacts.index(contact))
    print(f"Rimosso {removed[0]}")

def update_contact(old_contact, new_contact, contacts):
    contacts[contacts.index(old_contact)] = new_contact
    print(f"Aggiornato da {old_contact} a {new_contact}")

def search(contact, contacts):
    return contacts.index(contact) >= 0

def print_contacts(contacts):
    contacts.sort(key=lambda contact: contact[0])
    print(contacts)

contacts = [("Patrick Nicolosi", 3271220065), ("Katya Petralia", 3402621899), ("Pietro Nicolosi", 3349518776)]

add_contact(("Ciccio Parisi", 3807883842), contacts)
remove_contact(contacts[0], contacts)
update_contact(("Ciccio Parisi", 3807883842), ("Ciccio Pazzo", 3807883842), contacts)
print(search(contacts[0], contacts))
print_contacts(contacts)
`````

`````python
#Esercizio
#Fare un programma che calcola le statistiche di un testo quindi: contare il numero totale di parole, calcolare la frequenza di ogni parola, estrarre le parole uniche usando un set, mostrare le 5 parole piu frequenti

def n_words(s):
    return len(s.split())

def words_frequency(s):
    ws = s.split()
    fs = {}
    for w in ws:
        fs[w] = fs.get(w, 0) + 1
    return fs


s = "ciao come stai, io credo di non aver mai visto un testo cosi bello e ho visto e guardato tanto testo in vita mia. In ogni caso ti saluto e ti dico ancora ciao"

print(n_words(s))
print(words_frequency(s))


`````
